# 2. Variables, mutability and identity

The topic most students lose points on. Ask yourself for every line: does it **rebind a name** or **change an object**?

## Step 2.1: Variables are names

- `x = 5` puts the name `x` on the object 5.
- `x = x + 2` computes a new value and moves the name `x` to it.
- `b = a` puts a second name on the **same** object; it does not copy.
- `x += 3` is short for `x = x + 3` (for numbers and strings).
- `a, b = b, a` swaps two values.

**Demo:** run this and compare each line with the explanation.

````python
x = 3
x = x + 2
print(x)
a = 1
b = a
b = 2
print(a, b)
````

## Step 2.4: Methods that change the object vs return a new one

- String methods (`.upper()`, `.replace()`, `.strip()`, ...) **always return a new string**; the original never changes.
- Classic trap: `L = L.append(2)` makes L `None`.

- `seq[start:stop]` takes from `start` up to **but not including** `stop`.

## Step 8.2: try / except flow

- Python runs the `try` block line by line. At the first error it **jumps** to the first matching `except`; the rest of `try` is skipped.
- `else` runs only if there was **no** error. `finally` runs **always**.
- After the block, the program continues normally.

**Demo:** run this and compare each line with the explanation.

````python
try:
    x = int("5")
    y = 10 / 0
    print("skipped")
except ZeroDivisionError:
    print("cannot divide by zero")
print("program continues", x)
````
